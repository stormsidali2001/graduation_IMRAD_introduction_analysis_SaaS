# Confusion Matrices — Inference Only (Hugging Face)

Loads each saved BERT model and the dataset **directly from Hugging Face**, reproduces the exact same train/val/test split (`seed=42`) used during training, runs inference on the test set, and saves normalized confusion matrix PNGs.

**No training, no Google Drive required.** Run all cells top to bottom.

> **Note on split reproducibility:** The split uses `df.sample(frac=0.8, random_state=42)`, which depends on row order. If the Hugging Face dataset preserves the original CSV row order, this produces the identical test set used during training. Either way, the result is a valid evaluation on a representative held-out set.

## 2. Imports

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import tensorflow as tf
import tensorflow_hub as hub
import tensorflow_text          # required to deserialise TF Hub BERT layers
import keras
from keras.utils import to_categorical

from sklearn import preprocessing
from sklearn.metrics import confusion_matrix, classification_report
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from datasets import load_dataset
from huggingface_hub import from_pretrained_keras

print("TF:", tf.__version__)
tf.get_logger().setLevel('ERROR')
```

## 3. Load dataset from Hugging Face

```python
DATASET_ID = "stormsidali2001/IMRAD-introduction-sentences-moves-sub-moves-dataset"

hf_ds = load_dataset(DATASET_ID, split="train")
raw_df = hf_ds.to_pandas()
print("Loaded:", raw_df.shape)
print(raw_df.columns.tolist())
```

### Preprocess (mirrors every training notebook)

```python
# Convert label to int; -1 marks outliers added during annotation
raw_df['gemini_move'] = raw_df['move_sub_move_gemini'].apply(int)
raw_df['sentence']    = raw_df['sentence'].apply(lambda x: str(x).lower())

# Remove outliers
full_df = raw_df[raw_df['gemini_move'] != -1].copy()
print("After outlier removal:", full_df.shape)
print(full_df['gemini_move'].value_counts())
```

## 4. Shared helpers

```python
SEED  = 42
BATCH = 32

def make_split(df):
    """Exact replica of the split used in every training notebook."""
    train     = df.sample(frac=0.8, random_state=SEED)
    remainder = df.drop(train.index)
    val       = remainder.sample(frac=0.5, random_state=SEED)
    test      = remainder.drop(val.index)
    return train, val, test

def make_tf_dataset(data, label_col, n_classes):
    """Non-shuffled ordered dataset for aligned prediction."""
    texts  = data['sentence'].to_numpy()
    labels = to_categorical(data[label_col].to_numpy(), num_classes=n_classes)
    return tf.data.Dataset.from_tensor_slices((texts, labels)).batch(BATCH)

def make_text_dataset(data):
    """Text-only dataset for embedding extraction (no labels needed)."""
    texts = data['sentence'].to_numpy()
    return tf.data.Dataset.from_tensor_slices(texts).batch(BATCH)

def plot_and_save(cm, class_labels, title, cmap, filepath, figsize):
    fig, ax = plt.subplots(figsize=figsize)
    sns.heatmap(cm, annot=True, fmt='.3f', cmap=cmap,
                xticklabels=class_labels, yticklabels=class_labels,
                ax=ax, vmin=0, vmax=1)
    ax.set_title(title, fontweight='bold')
    ax.set_ylabel('True Label')
    ax.set_xlabel('Predicted Label')
    plt.tight_layout()
    plt.savefig(filepath, dpi=150, bbox_inches='tight')
    plt.show()
    print(f'Saved: {filepath}')

def get_embedding_model(model):
    """Return a sub-model that outputs the BERT pooled_output (768-dim [CLS] vector)."""
    for layer in model.layers:
        try:
            out = layer.output
            if isinstance(out, dict) and 'pooled_output' in out:
                return tf.keras.Model(inputs=model.input, outputs=out['pooled_output'])
        except Exception:
            pass
    # Fallback: layer before the final Dense classifier
    return tf.keras.Model(inputs=model.input, outputs=model.layers[-2].output)

def extract_embeddings(emb_model, data, max_samples=5000):
    """Extract BERT embeddings; cap at max_samples for t-SNE speed."""
    if len(data) > max_samples:
        data = data.sample(n=max_samples, random_state=SEED)
    ds = make_text_dataset(data)
    embs = emb_model.predict(ds)
    return embs, data

def plot_pca_tsne(embeddings, labels, label_names, title_prefix, pca_path, tsne_path):
    """Plot and save PCA and t-SNE of BERT embeddings side by side as two files."""
    n_classes = len(label_names)
    palette   = sns.color_palette('tab10', n_classes)

    # PCA
    coords_pca = PCA(n_components=2, random_state=SEED).fit_transform(embeddings)
    fig, ax = plt.subplots(figsize=(6, 5))
    for i, name in enumerate(label_names):
        mask = labels == i
        ax.scatter(coords_pca[mask, 0], coords_pca[mask, 1],
                   label=name, s=8, alpha=0.5, color=palette[i])
    ax.set_title(f'{title_prefix} — PCA (BERT embeddings)', fontweight='bold')
    ax.set_xlabel('PC 1'); ax.set_ylabel('PC 2')
    ax.legend(markerscale=3, fontsize=8, loc='best')
    plt.tight_layout()
    plt.savefig(pca_path, dpi=150, bbox_inches='tight')
    plt.show()
    print(f'Saved: {pca_path}')

    # t-SNE
    coords_tsne = TSNE(n_components=2, random_state=SEED, perplexity=30,
                       n_iter=1000, init='pca').fit_transform(embeddings)
    fig, ax = plt.subplots(figsize=(6, 5))
    for i, name in enumerate(label_names):
        mask = labels == i
        ax.scatter(coords_tsne[mask, 0], coords_tsne[mask, 1],
                   label=name, s=8, alpha=0.5, color=palette[i])
    ax.set_title(f'{title_prefix} — t-SNE (BERT embeddings)', fontweight='bold')
    ax.set_xlabel('Dim 1'); ax.set_ylabel('Dim 2')
    ax.legend(markerscale=3, fontsize=8, loc='best')
    plt.tight_layout()
    plt.savefig(tsne_path, dpi=150, bbox_inches='tight')
    plt.show()
    print(f'Saved: {tsne_path}')
```

## 5. Model 1 — Overall move classifier (3 classes)

```python
_, _, test1 = make_split(full_df)
print(f"Test set: {len(test1)} sentences")
test_ds1 = make_tf_dataset(test1, 'gemini_move', n_classes=3)

model1 = from_pretrained_keras("stormsidali2001/IMRAD_introduction_moves_classifier")
print("Model 1 loaded.")

y_pred1 = np.argmax(model1.predict(test_ds1), axis=1)
y_true1 = test1['gemini_move'].values

print(classification_report(y_true1, y_pred1,
      target_names=['Move 0', 'Move 1', 'Move 2']))

cm1 = confusion_matrix(y_true1, y_pred1, normalize='true')
plot_and_save(
    cm1,
    ['Move 0\n(Research Territory)', 'Move 1\n(Establish Niche)', 'Move 2\n(Occupy Niche)'],
    'Model 1: Overall Move Classifier\nNormalized Confusion Matrix',
    'Blues', 'confusion_matrix_model1.png', (6, 5))
```

## 6. Model 2 — Move 0 sub-move specialist (2 classes)

```python
df2 = full_df[full_df['gemini_move'] == 0].copy()
df2 = df2.sort_values(by='move_sub_move_gemini', ascending=True)
le2 = preprocessing.LabelEncoder()
df2['encoded'] = le2.fit_transform(df2['move_sub_move_gemini'])
print("Classes:", le2.classes_)   # [0.0, 0.1]

_, _, test2 = make_split(df2)
print(f"Test set: {len(test2)} sentences")
test_ds2 = make_tf_dataset(test2, 'encoded', n_classes=len(le2.classes_))

model2 = from_pretrained_keras("stormsidali2001/IMRAD-introduction-move-zero-sub-moves-classifier")
print("Model 2 loaded.")

y_pred2 = np.argmax(model2.predict(test_ds2), axis=1)
y_true2 = test2['encoded'].values

print(classification_report(y_true2, y_pred2,
      target_names=[str(c) for c in le2.classes_]))

cm2 = confusion_matrix(y_true2, y_pred2, normalize='true')
plot_and_save(
    cm2,
    ['0.0: Establish\nimportance/relevance', '0.1: Review\nprior research'],
    'Model 2: Move 0 Sub-move Specialist\nNormalized Confusion Matrix',
    'Greens', 'confusion_matrix_model2_move0.png', (5, 4))
```

## 7. Model 3 — Move 1 sub-move specialist (4 classes)

```python
df3 = full_df[full_df['gemini_move'] == 1].copy()
df3 = df3.sort_values(by='move_sub_move_gemini', ascending=True)
le3 = preprocessing.LabelEncoder()
df3['encoded'] = le3.fit_transform(df3['move_sub_move_gemini'])
print("Classes:", le3.classes_)   # [1.0, 1.1, 1.2, 1.3]

_, _, test3 = make_split(df3)
print(f"Test set: {len(test3)} sentences")
test_ds3 = make_tf_dataset(test3, 'encoded', n_classes=len(le3.classes_))

model3 = from_pretrained_keras("stormsidali2001/IMRAD-introduction-move-one-sub-moves-classifier")
print("Model 3 loaded.")

y_pred3 = np.argmax(model3.predict(test_ds3), axis=1)
y_true3 = test3['encoded'].values

print(classification_report(y_true3, y_pred3,
      target_names=[str(c) for c in le3.classes_]))

cm3 = confusion_matrix(y_true3, y_pred3, normalize='true')
plot_and_save(
    cm3,
    ['1.0: Claim flaw\nin prior work', '1.1: Highlight\na gap',
     '1.2: Raise unclear\nquestion', '1.3: Extend\nprior research'],
    'Model 3: Move 1 Sub-move Specialist\nNormalized Confusion Matrix',
    'Blues', 'confusion_matrix_model3_move1.png', (7, 6))
```

## 8. Model 4 — Move 2 sub-move specialist (5 classes)

```python
df4 = full_df[full_df['gemini_move'] == 2].copy()
df4 = df4.sort_values(by='move_sub_move_gemini', ascending=True)
le4 = preprocessing.LabelEncoder()
df4['encoded'] = le4.fit_transform(df4['move_sub_move_gemini'])
print("Classes:", le4.classes_)   # [2.0, 2.1, 2.2, 2.3, 2.4]

_, _, test4 = make_split(df4)
print(f"Test set: {len(test4)} sentences")
test_ds4 = make_tf_dataset(test4, 'encoded', n_classes=len(le4.classes_))

model4 = from_pretrained_keras("stormsidali2001/IMRAD-introduction-move-two-sub-moves-classifier")
print("Model 4 loaded.")

y_pred4 = np.argmax(model4.predict(test_ds4), axis=1)
y_true4 = test4['encoded'].values

print(classification_report(y_true4, y_pred4,
      target_names=[str(c) for c in le4.classes_]))

cm4 = confusion_matrix(y_true4, y_pred4, normalize='true')
plot_and_save(
    cm4,
    ['2.0: State\npurpose/aim', '2.1: State\nhypothesis',
     '2.2: Share\nfindings', '2.3: Elaborate\nvalue', '2.4: Outline\nstructure'],
    'Model 4: Move 2 Sub-move Specialist\nNormalized Confusion Matrix',
    'Blues', 'confusion_matrix_model4_move2.png', (8, 7))
```

## 10. BERT Embedding Visualisations (PCA + t-SNE)

Extracts the 768-dimensional BERT `[CLS]` pooled output from each model's encoder and projects it to 2D with PCA and t-SNE. Capped at 5,000 samples for t-SNE speed; PCA uses the full test set.

### Model 1 — Overall move classifier (3 classes)

```python
emb_model1 = get_embedding_model(model1)
embs1, sampled1 = extract_embeddings(emb_model1, test1)
labels1 = sampled1['gemini_move'].values

plot_pca_tsne(
    embs1, labels1,
    label_names=['Move 0: Research Territory',
                 'Move 1: Establish Niche',
                 'Move 2: Occupy Niche'],
    title_prefix='Model 1 (3 moves)',
    pca_path='pca_bert_model1.png',
    tsne_path='tsne_bert_model1.png'
)
```
